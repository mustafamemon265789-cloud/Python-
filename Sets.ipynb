# **Frozen Set**

```python
frozenset1=frozenset([1,2,3,4])
frozenset2=frozenset([3,4,5,6])
frozenset3=frozenset([1,2])
print(f"Frozenset1 =",frozenset1)
print(f"Frozenset2 =",frozenset2)
print(f"Frozenset3 =",frozenset3)
print("\n----------\n")

differenceset=frozenset1.difference(frozenset2)
print(f"difference() =,{differenceset}")

intersectionset=frozenset1.intersection(frozenset2)
print(f"intersection() =,{intersectionset}")

unionset=frozenset1.union(frozenset2)
print(f"union() =,{unionset}")

symmetric_difference_set=frozenset1.symmetric_difference(frozenset2)
print(f"symmetric_difference() =,{symmetric_difference_set}")
```
